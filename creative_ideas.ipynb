# Creative Ideas for Everyday Objects

What else could ordinary objects do? This notebook asks you for objects one at a time and uses the OpenAI API to suggest three unusual, practical uses for each one.

Run the cells from top to bottom. You will need an OpenAI API key. The key is requested privately while the notebook is running; do not paste it into a code cell.

```python
# Install the official OpenAI Python SDK in this Colab session.
%pip -q install openai
```

```python
from getpass import getpass
from openai import OpenAI

# Enter your key when prompted. getpass hides what you type.
api_key = getpass("OpenAI API key (input is hidden): ")
client = OpenAI(api_key=api_key)

# Add an object, like a mug or a paper clip.
# Keep asking for objects until you type quit.
while True:
    object_name = input("Name an everyday object (or type 'quit' to stop): ").strip()
    if object_name.lower() == "quit":
        print("Goodbye!")
        break
    if not object_name:
        print("Please enter an object name.")
        continue

    # Ask for three unusual, practical, and safe ideas.
    response = client.responses.create(
        model="gpt-6-luna",
        input=(
            f"Suggest exactly THREE unusual, practical, and safe uses for {object_name!r}. "
            "List them as 1, 2, and 3, with a short explanation for each."
        ),
    )

    # Show the ideas, then ask for another object.
    print(response.output_text)
    print()
```
