# Creative Ideas for Everyday Objects

What else could an ordinary object do? This notebook asks you for an everyday object and uses the OpenAI API to suggest one unusual, practical use.

Run the cells from top to bottom. You will need an OpenAI API key. The key is requested privately while the notebook is running; do not paste it into a code cell.

```python
# Install the official OpenAI Python SDK in this Colab session.
%pip -q install openai
```

```python
from getpass import getpass
from openai import OpenAI

# Enter your key when prompted. getpass hides what you type.
api_key = getpass("OpenAI API key (input is hidden): ")
client = OpenAI(api_key=api_key)

# Tell us an ordinary object, like a mug or a paper clip.
object_name = input("Name an everyday object: ").strip()
if not object_name:
    raise ValueError("Please enter an everyday object and run this cell again.")

# Ask for exactly one unusual, practical, and safe idea.
response = client.responses.create(
    model="gpt-6-luna",
    input=(
        f"Suggest exactly ONE unusual, practical, and safe use for {object_name!r}. "
        "Explain the idea in 2-3 friendly sentences."
    ),
)

# Show the suggestion without displaying the API key.
print(response.output_text)
```
